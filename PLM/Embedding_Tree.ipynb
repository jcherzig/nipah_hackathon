```python
import json
import pandas as pd

with open('/Users/kieran.lamb/Github/nipah_hackathon/all_sequences_and_embeddings.json', 'r') as f:
    data = json.load(f)

data = pd.DataFrame(data).T.reset_index().rename(columns={'index':'sequence_id'})


metadata = pd.read_csv('/Users/kieran.lamb/Github/nipah_hackathon/wide_evaluation_data.csv')
metadata.drop(columns=['sequence'], inplace=True)
metadata['sequence_id'] = [ data[data.sequence_id.str.contains(seq_clean) == True].sequence_id.iloc[0]  for seq_clean in metadata.id]



data = pd.merge(data,metadata, left_on='sequence_id',right_on='sequence_id')
data
```

```python
import io
import numpy as np
from scipy.cluster import hierarchy
from Bio import Phylo
import scipy


# Calculate pairwise Euclidean distances (condensed distance matrix)
from scipy.spatial.distance import cosine, pdist

dist_matrix = pdist(np.array([e for e in data.embeddings]), metric='cosine')
dist_matrix = scipy.spatial.distance.squareform(dist_matrix)


# 2. Perform Hierarchical Clustering (Linkage Matrix)
Z = hierarchy.linkage(dist_matrix, method='ward')

# 3. Recursive function to convert Scipy Dendrogram to Newick format
def to_newick(node, newick, parent_dist, leaf_names):
    if node.is_leaf():
        return f"{leaf_names[node.id]}:{parent_dist - node.dist:.4f}{newick}"
    else:
        if len(newick) > 0:
            newick = f"):{parent_dist - node.dist:.4f}{newick}"
        else:
            newick = ");"
        
        # Traverse right and left branches
        newick = to_newick(node.get_left(), newick, node.dist, leaf_names)
        newick = to_newick(node.get_right(), f",{newick}", node.dist, leaf_names)
        return f"({newick}"
    


# 4. Create the Newick string
tree_root = hierarchy.to_tree(Z, rd=False)
newick_str = to_newick(tree_root, "", tree_root.dist, data.id)

newick_str

# # 5. Load the Newick string into a Biopython Phylo Tree
tree_handle = io.StringIO(newick_str)
phylo_tree = Phylo.read(tree_handle, "newick")

Phylo.write(phylo_tree, "embedding_tree.nwk", "newick")

Phylo.draw_ascii(phylo_tree)
```
